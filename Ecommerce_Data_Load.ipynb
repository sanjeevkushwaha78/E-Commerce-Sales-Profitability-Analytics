```python
# Create Analytics Dataset

df=orders.merge(
    customers,
    on='CustomerID',
    how="left"
)
```

```python
## Add OrderDetals

df=df.merge(
    order_details,
    on="OrderID",
    how='inner'
)
```

```python
# Add Product Information

df=df.merge(
    products,
    on="ProductID",
    how="left"
)
```

```python
df.shape
```

```python
df.head()
```

```python
df['Year']=df["OrderDate"].dt.year
df['Month']=df["OrderDate"].dt.month
df['MonthName']=df["OrderDate"].dt.month_name()
df['YearMonth']=df["OrderDate"].dt.to_period('M').astype(str)
```

```python
## Profit Margin

df["ProfitMarginPct"]=(
    df["Profit"]/df["SalesAmount"]
)*100
```
