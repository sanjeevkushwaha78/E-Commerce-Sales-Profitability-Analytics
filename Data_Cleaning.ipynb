```python
import pandas as pd
import numpy as np
import seaborn as sns
import  matplotlib.pyplot as plt

from scipy.stats import ttest_ind
```

```python
df=pd.read_csv('ecommerce_clean_data.csv')
```

```python
df.head()
df.shape
```

```python
## Fillter Delivered Orders
test_df=df[df["OrderStatus"]=="Delivered"].copy()
```

```python
test_df
```

```python
test_df.shape
```

```python
## Understand Discount Data

test_df["DiscountPct"].describe()
```

```python
# Step1 Order-Level aggregation

order_level=(
    test_df.groupby("OrderID")
    .agg(
        orderValue=("SalesAmount", "sum"),
        DiscountPct=("DiscountPct", "mean")
    )
    .reset_index()
)
```

```python
order_level
```

```python
discount_group=order_level[order_level["DiscountPct"] > 0] ["orderValue"]
```

```python
discount_group
```

```python
no_discount_group=order_level[order_level["DiscountPct"]==0]["orderValue"]
```

```python
no_discount_group
```

```python
# Group Size


print('Discount Orders:', len(discount_group))
print('No Discount Orders:', len(no_discount_group))
```

```python
# Group Mean

print('Discount AOV:',
      discount_group.mean()
     )
print("No Discount AOV :", no_discount_group.mean()
     )
```

```python
# Group Median

print('Discount MOV:', discount_group.median()
     )
print('No Discount MOV:', no_discount_group.median()
     )
```

```python
# Visual Comparison

Comparison_df=pd.DataFrame({
    'AOV': pd.concat(
        [discount_group, no_discount_group],
        ignore_index=True),
    'Group':(
        ['Discount']* len(discount_group)
        +
        ['No Discount'] * len(no_discount_group)
        )})
```

```python
Comparison_df
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=Comparison_df,
    x="Group",
    y="AOV"
)

plt.title("AOV: Discount vs No Discount")
plt.xlabel("Group")
plt.ylabel("Average Order Value")

plt.show()
```

```python
# Independent t-test

t_stat, p_value= ttest_ind(
    discount_group,
    no_discount_group,
    equal_var=False)
```

```python
## Result

print("T-Statistic:" , t_stat)
print("P-Value:" ,p_value)
```

```python
# Decision Rule

alpha=0.05
if p_value<alpha:
    print('Reject H0')
else:
    print('Fail to Reject Ho')
    
```

```python
from scipy.stats import ttest_ind

result= ttest_ind(
    discount_group,
    no_discount_group,
    equal_var=False
)
print(result)
```

```python
result.confidence_interval()
```

```python
## Effect Difference

aov_difference=(
    discount_group.mean()-no_discount_group.mean())
aov_difference
```

```python
## Final Hypothesis Testing Summary

hypothesis_result = pd.DataFrame({
    "Metric": [
        "Discount AOV",
        "No Discount AOV",
        "Difference",
        "T-Statistic",
        "p_value",
        "Alpha"
    ],
    "Value": [
        discount_group.mean(),
        no_discount_group.mean(),
        aov_difference,
        t_stat,
        p_value,
        alpha
    ]
})

hypothesis_result
```
